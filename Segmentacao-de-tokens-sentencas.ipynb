```python
frases_nltk = nltk.tokenize.sent_tokenize(texto, language='portuguese')
```

TOKENIZANDO MANUALMENTE

```python
frases_nltk
```

```python
frases_tokenizadas = []

for frase in frases_nltk:
    frase = frase.replace(".", " . ")
    frase = frase.replace(",", " , ")
    frase = frase.replace(":", " : ")
    frase = frase.replace("(", " ( ")
    frase = frase.replace(")", " ) ")
    frase = frase.replace("-", " _ ")
    frase = frase.replace("«", " « ")
    frase = frase.replace("»", " » ")
    frase = frase.replace("   ", " ")
    frase = frase.strip()
    frases_tokenizadas.append(frase.split())

```

TOKENIZANDO USANDO NLTK

```python
frases_nltk
```

```python
frases_tokenizadas_nltk = []

for frase in frases_nltk:
    frase = nltk.tokenize.word_tokenize(frase)
    frases_tokenizadas_nltk.append(frase)
```

```python
frases_tokenizadas_nltk = [nltk.word_tokenize(frase) for frase in frases_nltk]
frases_sem_stopwords = []

for frase in frases_tokenizadas_nltk:
    frase_limpa = []
    for token in frase:
        if token.lower() not in stopwords:
            frase_limpa.append(token)
    frases_sem_stopwords.append(frase_limpa)
```

REMOVENDO STOPWORDS USANDO NLTK

```python

import string
string.punctuation
```

```python

nltk.download("stopwords")
```

```python
nltk.corpus.stopwords.words('portuguese')
```

```python

frases_sem_stopwords_nltk = []

for frase in frases_tokenizadas_nltk:
    frase_sem_stopwords = []
    for token in frase:
        if not token.lower() in nltk.corpus.stopwords.words('portuguese') and not token in string.punctuation:
            frase_sem_stopwords.append(token)
    frases_sem_stopwords_nltk.append(frase_sem_stopwords)
```
