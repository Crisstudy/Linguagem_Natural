Dataset e-commerce

```python
import pandas as pd

url = "https://raw.githubusercontent.com/americanas-tech/b2w-reviews01/master/B2W-Reviews01.csv"
dataset_ecommerce = pd.read_csv(url)
```

```python
dataset_ecommerce = [texto for texto in dataset_ecommerce['review_text'] if isinstance(texto, str)]
```

PRE-PROCESSAMENTO DOS TEXTOS 

```python
dataset_academico
```

```python
documentos = {}
```

```python
documentos['academicos'] = " ".join(dataset_academico)
```

```python
documentos['academicos'][:5000]
```

```python
documentos['ecommerce'] = " ".join(dataset_ecommerce)
```

```python
documentos['ecommerce'][:5000]
```

```python
def pre_process(text):
    text = text.replace(",", " ")
    text = text.replace(".", " ")
    text = text.replace("!", " ")
    text = text.replace("?", " ")
    text = text.replace("(", " ")
    text = text.replace(")", " ")
    text = text.lower()
    text = text.split()
    return text
```

```python
documentos['academicos'] = pre_process(documentos['academicos'])
documentos['ecommerce'] = pre_process(documentos['ecommerce'])
```

```python
documentos['ecommerce'][:20]
```

FREQUENCIA PROPORCIONAL

```python
documentos['academicos']
```

```python
qtd_tokens = {
    'academico': len(documentos['academicos']),
    'ecommerce': len(documentos['ecommerce'])
}
```

```python
qtd_tokens
```

```python
contador_frequencia
```

```python
freq_proporcional = {
    'academico': {},
    'ecommerce': {}
}
```

```python
# Um só loop para os dois gêneros
for genero in contador_frequencia:
    for token in contador_frequencia[genero]:
        freq_proporcional[genero][token] = contador_frequencia[genero][token] / qtd_tokens[genero]
```

```python
freq_proporcional['academico']
```

```python
freq_proporcional['ecommerce']
```
